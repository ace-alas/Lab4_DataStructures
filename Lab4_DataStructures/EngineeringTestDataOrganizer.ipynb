```python
# Student Identity
LAST_NAME = "Siruno"
STUDENT_ID = "TUPM-26-7958"
FAVORITE_ARTIST = "BINI"
SEED_NUM = 7

# Derive computational parameters from the fixed inputs
seed_digit = int(STUDENT_ID[-1])
id_checksum = sum(int(d) for d in STUDENT_ID if d.isdigit())
vector_dim = len(LAST_NAME)
artist_positions = [ord(c) - 64 for c in FAVORITE_ARTIST if c.isalpha()]

# Encapsulate state into a Dictionary
sys_config = {
    "operator": LAST_NAME,
    "artist": FAVORITE_ARTIST,
    "authid": seed_digit,
    "baseseed": SEED_NUM,
    "checksum": id_checksum,
    "vectordim": vector_dim,
    "positions": artist_positions,
    "status": "INITIALIZED"
}

print("=== SYSTEM CONFIGURATION ===")
for key, value in sys_config.items():
    print(f"{key.upper()}: {value}")
```

```python
base = sys_config["baseseed"]
positions = sys_config["positions"]

sources = [f"{sys_config['operator']}_T{i}" for i in range(1, 6)]
categories = ["TEMPERATURE", "VOLTAGE", "CURRENT", "PRESSURE", "SPEED"]

# Build one record per source/category pair, with extra measurements
# under the first two categories
test_records = []
for row, src in enumerate(sources):
    for col, cat in enumerate(categories):
        val = base * (row + 1) + positions[(row + col) % len(positions)]
        test_records.append((src, cat, val))
        if col < 2:                       # additional measurements
            test_records.append((src, cat, val + base))

# One repeated test-source/category record
test_records.append(test_records[0])

print("=== GENERATED TEST DATA ===")
for rec in test_records:
    print(rec)
print(f"\nTotal generated records: {len(test_records)}")
```

```python
organized = {}
for src, cat, val in test_records:
    organized.setdefault(src, {}).setdefault(cat, []).append(val)

print("=== ORGANIZED TEST DATA ===")
for src in sorted(organized):
    print(f"\nSource: {src}")
    for cat in sorted(organized[src]):
        print(f"  {cat:<12} -> {organized[src][cat]}")
```

```python
record_set = set(test_records)
repeated = [rec for rec in record_set if test_records.count(rec) > 1]

print("=== REPEATED / DISTINCT DATA ===")
print(f"Total records   : {len(test_records)}")
print(f"Distinct records: {len(record_set)}")
print(f"Repeated records: {len(repeated)}")
for rec in repeated:
    print(f"  repeated {rec} (x{test_records.count(rec)})")
```

```python
all_values = [val for src, cat, val in test_records]

num_summary = {
    "count": len(all_values),
    "minimum": min(all_values),
    "maximum": max(all_values),
    "total": sum(all_values)
}
num_summary["average"] = round(num_summary["total"] / num_summary["count"], 2)

cat_summary = {}
for src, cat, val in test_records:
    cat_summary[cat] = cat_summary.get(cat, 0) + 1

print("=== NUMERICAL SUMMARY ===")
for key, value in num_summary.items():
    print(f"{key.capitalize():<10}: {value}")

print("\n=== CATEGORICAL SUMMARY ===")
for cat in sorted(cat_summary):
    print(f"{cat:<12}: {cat_summary[cat]} record(s)")
```

```python
# Average value per source, highest first
source_averages = {}
for src, cat, val in test_records:
    source_averages.setdefault(src, []).append(val)
for src in source_averages:
    readings = source_averages[src]
    source_averages[src] = round(sum(readings) / len(readings), 2)

ordered_results = sorted(source_averages.items(), key=lambda item: item[1],
                         reverse=True)

print("=== ORDERED RESULTS (Average per Source) ===")
for rank, (src, avg) in enumerate(ordered_results, start=1):
    print(f"{rank}. {src:<20} average = {avg}")

print("\n=== FINAL SUMMARY ===")
print(f"Operator        : {sys_config['operator']}")
print(f"Sources         : {len(sources)}")
print(f"Categories      : {len(categories)}")
print(f"Records         : {len(test_records)}")
print(f"Distinct records: {len(record_set)}")
print(f"Top source      : {ordered_results[0][0]} ({ordered_results[0][1]})")
```
